### **1. Word2Vec Skip-gram 모델 구현**

```python
#라이브러리 임포트
import torch
import torch.nn as nn
import torch.optim as optim
import numpy as np
from collections import defaultdict
from sklearn.manifold import TSNE
import matplotlib.pyplot as plt
```

```python
#corpus(말뭉치) 만들기 -> 직접 새로운 말뭉치를 만들어봐도 좋습니다!
corpus = [
    "he likes cat", "she likes dog", "dog barks", "cat meows",
    "he walks the dog", "she feeds the cat", "the cat sleeps on the sofa",
    "the dog plays in the yard", "they both like animals", "animals are friends",
    "cats are cute", "dogs are the best", "she plays with the cat", "she is not afraid of animals"
]
```

```python
# 단어 사전 구축
word_list = set(" ".join(corpus).split()) # 모든 문장을 띄어쓰기로 분리
word_to_ix = {word: i for i, word in enumerate(word_list)}  # 단어 -> 인덱스
ix_to_word = {i: word for word, i in word_to_ix.items()}  # 인덱스 -> 단어
vocab_size = len(word_to_ix)  # 어휘 사전의 크기
```

```python
# context-target 쌍 만들기
def make_context_target(corpus, window_size=1):
    pairs = []
    for sentence in corpus:
        words = sentence.split()
        for i, word in enumerate(words):
            for j in range(max(i - window_size, 0), min(i + window_size + 1, len(words))):
                if i != j:
                    pairs.append((word, words[j]))  # 중심 단어 - 주변 단어 쌍 저장
    return pairs

pairs = make_context_target(corpus)

# 단어를 텐서 인덱스로 변환
def word_to_tensor(word):
    return torch.tensor([word_to_ix[word]], dtype=torch.long)
```

```python
#VanillaSkipgram 정의
class VanillaSkipgram(nn.Module):
    def __init__(self, vocab_size, embedding_dim):
        super().__init__()
        self.embedding = nn.Embedding(vocab_size, embedding_dim)  # 입력 단어 임베딩 벡터 생성
        self.linear = nn.Linear(embedding_dim, vocab_size)  # 출력층: 단어 사전 크기의 로짓 출력

    def forward(self, input_ids):
        embeddings = self.embedding(input_ids)  # (1, ) -> (1, embedding_dim)
        output = self.linear(embeddings)  # (1, embedding_dim) -> (1, vocab_size)
        return output
```

```python
#Cross Entropy Loss, Adam Optimizer(learning rate = 0.01)를 사용하세요
embedding_dim = 10
model = VanillaSkipgram(vocab_size, embedding_dim)
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=0.01)
```

```python
for epoch in range(200):
    total_loss = 0
    for center, context in pairs:
        center_tensor = word_to_tensor(center)  # 중심 단어 입력
        context_tensor = word_to_tensor(context)  # 주변 단어 (정답)
        optimizer.zero_grad()
        outputs = model(center_tensor)  # (1, vocab_size)
        loss = criterion(outputs, context_tensor) # 실제 주변 단어와 비교한 손실
        loss.backward()
        optimizer.step()
        total_loss += loss.item()
    if (epoch + 1) % 20 == 0:
        print(f"Epoch {epoch+1}, Loss: {total_loss:.4f}")
```

```python
# 결과 확인
word = "dog"  #다른 단어로 바꿔서 확인해볼 수도 있습니다
word_tensor = word_to_tensor(word)  # 단어 인덱스 -> 텐서
embedding = model.embedding(word_tensor)  # 해당 단어의 임베딩 벡터 추출
print(f"'{word}'의 임베딩 벡터:")
print(embedding)
```

```python
#시각화하기
embedding_matrix = model.embedding.weight.data.cpu().numpy()

words = list(word_to_ix.keys())[:20]
vectors = np.array([embedding_matrix[word_to_ix[word]] for word in words])

tsne = TSNE(n_components=2, perplexity=5, max_iter=300, random_state=0)
reduced = tsne.fit_transform(vectors)
```

```python
plt.figure(figsize=(8, 6))
for i, word in enumerate(words):
    x, y = reduced[i]
    plt.scatter(x, y)
    plt.text(x + 0.01, y + 0.01, word, fontsize=10)
plt.title("Word Embeddings")
plt.grid(True)
plt.show()
```

### **2. RNN을 활용한 문장 분류 모델 구현**

```python
import torch.nn.functional as F
```

```python
train_data = [
    ("I love this movie", 1),
    ("This film is terrible", 0),
    ("I enjoyed it", 1),
    ("Not good", 0),
    ("This movie was fantastic and inspiring", 1),
    ("Absolutely horrible experience", 0),
    ("What a wonderful story and cast", 1),
    ("The plot was dull and boring", 0),
    ("A truly uplifting journey", 1),
    ("It was a waste of time", 0),
    ("I think they could have done better", 0),
    ("It was really enjoyable", 1),
    ("This became my favorite movie", 1),
    ("I recommended this to all of my friends", 1),
    ("My son could film a better movie than this", 0),
    ("It was underwhelming", 0)
]
```

```python
# 단어 사전
word_set = set()
for sent, _ in train_data:
    for word in sent.lower().split():
        word_set.add(word)

word_to_ix = {word: i+1 for i, word in enumerate(word_set)}  # 0: padding
vocab_size = len(word_to_ix) + 1

# 문장을 숫자 인덱스 시퀀스로 인코딩
def encode_sentence(sent, max_len=10):
    tokens = [word_to_ix.get(word, 0) for word in sent.lower().split()]
    tokens += [0] * (max_len - len(tokens))
    return torch.tensor(tokens[:max_len])

# X: 입력 텐서, y: 정답 텐서
X = torch.stack([encode_sentence(s) for s, _ in train_data])  # (batch_size, seq_len)
y = torch.tensor([label for _, label in train_data])  # (batch_size, )
```

```python
class RNNClassifier(nn.Module):
    def __init__(self, vocab_size, embedding_dim, hidden_dim):
        super().__init__()
        self.embedding = nn.Embedding(vocab_size, embedding_dim, padding_idx=0) # 단어 -> 임베딩 벡터
        self.rnn = nn.RNN(embedding_dim, hidden_dim, batch_first=True)  # RNN 셀 정의
        self.fc = nn.Linear(hidden_dim, 2)  # 최종 분류

    def forward(self, inputs):
        embeddings = self.embedding(inputs)
        output, hidden = self.rnn(embeddings)
        return self.fc(hidden.squeeze(0))
```

```python
model = RNNClassifier(vocab_size, embedding_dim=16, hidden_dim=32)
loss_fn = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=0.01)

for epoch in range(30):
    optimizer.zero_grad()
    outputs = model(X)  # 입력 시퀀스를 통해 로짓 계산
    loss = loss_fn(outputs, y)  # 정답과 비교하여 손실 계산
    loss.backward()
    optimizer.step()
    if (epoch + 1) % 10 == 0:
        print(f"Epoch {epoch+1}, Loss: {loss.item():.4f}")

preds = torch.argmax(model(X), dim=1)
print("예측 결과:", preds.tolist())
print("실제 정답:", y.tolist())
```

### **3. CNN을 활용한 문장 분류 모델 구현**

```python
#Conv1d로 정의하세요
class CNNClassifier(nn.Module):
    def __init__(self, vocab_size, embedding_dim, num_filters, kernel_size):
        super().__init__()

        self.embedding = nn.Embedding(vocab_size, embedding_dim, padding_idx=0) # 단어 인덱스 -> 밀집 벡터

        self.conv = nn.Conv1d(in_channels=embedding_dim, # 임베딩 차원(입력 피처 수)
                              out_channels=num_filters, # 필터 수
                              kernel_size=kernel_size)  # 필터 크기
        self.pool = nn.AdaptiveMaxPool1d(1) # 타임축에서 max 값 1개만 추출
        self.fc = nn.Linear(num_filters, 2) # 최종 분류기

    def forward(self, x):
        x = self.embedding(x).permute(0, 2, 1)  #(B, E, T). 각 단어 인덱스 -> 임베딩 벡터 -> Conv1d 입력 형태
        x = F.relu(self.conv(x))                #(B, F, T-k+1). 필터 적용, ReLU
        x = self.pool(x).squeeze(2)             #(B, F). 타임축 최대값 추출
        return self.fc(x) # 클래스별 점수 반환
```

```python
model = CNNClassifier(vocab_size, embedding_dim=16, num_filters=32, kernel_size=3)
loss_fn = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=0.01)

for epoch in range(30):
    optimizer.zero_grad()
    outputs = model(X)  # 순전파
    loss = loss_fn(outputs, y)  # 손실 계산
    loss.backward() # 역전파
    optimizer.step()
    if (epoch + 1) % 10 == 0:
        print(f"Epoch {epoch+1}, Loss: {loss.item():.4f}")

preds = torch.argmax(model(X), dim=1)
print("예측 결과:", preds.tolist())
print("실제 정답:", y.tolist())
```
